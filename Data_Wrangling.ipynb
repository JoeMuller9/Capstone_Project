**IDENTIFYING AND REMOVING INCONSISTENCIES**

```python
# Identifying countries in countries column in need of standardising 
countries = sorted(df['Country'].dropna().unique())
print(f"Total unique countries: {len(countries)}")

flag_keywords = ['Republic', 'Federation', 'Democratic', 'Islamic', 'Bolivarian', "People's", '...', ',']
candidates = [c for c in countries if any(k in c for k in flag_keywords)]
print(f"Candidates needing standardization: {len(candidates)}")
candidates
```

```python
# Standardising countries column.
country_map = {
    'Congo, Republic of the...': 'Republic of the Congo',
    "Democratic People's Republic of Korea": 'North Korea',
    'Iran, Islamic Republic of...': 'Iran',
    "Lao People's Democratic Republic": 'Laos',
    'Micronesia, Federated States of...': 'Micronesia',
    'Republic of Korea': 'South Korea',
    'Republic of Moldova': 'Moldova',
    'Republic of North Macedonia': 'North Macedonia',
    'Russian Federation': 'Russia',
    'Syrian Arab Republic': 'Syria',
    'United Republic of Tanzania': 'Tanzania',
    'Venezuela, Bolivarian Republic of...': 'Venezuela',
}

df['Country'] = df['Country'].replace(country_map)

print(f"Unique countries after standardization: {df['Country'].nunique()}")
df['Country'].value_counts()
```

```python
# Standardising EdLevel column
edlevel_counts = df['EdLevel'].value_counts(dropna=False)
print(f"Number of unique EdLevel values: {df['EdLevel'].nunique()}")

df['EdLevel'] = df['EdLevel'].str.replace(r'\s*\(.*?\)', '', regex=True).str.strip()

df['EdLevel'].value_counts(dropna=False)
```

**Encoding Categorical Variables**

```python
# Encode the employment column using one-hot encoding
# Analyse different results in employment column
df['Employment'].value_counts()
```

```python
# Encode the Employment column
employment_dummies = df['Employment'].str.get_dummies(sep=';')
employment_dummies = employment_dummies.add_prefix('Employment_')

df = pd.concat([df, employment_dummies], axis=1)

print(f"New columns created: {employment_dummies.shape[1]}")
employment_dummies.columns.tolist()

# Sanity check
sample = df[df['Employment'] == 'Employed, full-time;Independent contractor, freelancer, or self-employed'].iloc[0]
sample[[c for c in employment_dummies.columns]]
```

**REMOVING MISSING VALUES**

```python
# Identifying columns with highest amounts of missing values

missing_data = df.isnull()
missing_data_counts = missing_data.sum().sort_values(ascending=False)
missing_data_counts
```

```python
# Replacing missing values in numeric columns with the mean values
df = df.fillna(df.mean(numeric_only=True))
```

```python
# Replacing missing values in categoric columns with the mode values
categorical_cols = df.select_dtypes(include='object').columns

df[categorical_cols] = df[categorical_cols].fillna(df[categorical_cols].mode().iloc[0])
```

```python
# Final check to see if any missing data remains

missing_data = df.isnull()
missing_data_counts = missing_data.sum().sort_values(ascending=False)

for i in missing_data_counts:
    if i > 0:
        print(i,'still has missing data')

else:
    print('there is no missing entries in the data')
```

**Feature Scaling and Transformation**

```python
# Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.
df['ConvertedCompYearly_MinMax'] = (df['ConvertedCompYearly'] - df['ConvertedCompYearly'].min()) / (df['ConvertedCompYearly'].max() - df['ConvertedCompYearly'].min())
df['ConvertedCompYearly_MinMax'] .value_counts()
```

```python
# Log-transform the ConvertedCompYearly column to reduce skewness.

df['ConvertedCompYearly_Log'] = np.log1p(df['ConvertedCompYearly'])

print("Original skew:", df['ConvertedCompYearly'].skew())
print("Log-transformed skew:", df['ConvertedCompYearly_Log'].skew())
```

**Feature Engineering**

```python
# Create a new column `ExperienceLevel` based on the `YearsCodePro` column:
df['YearsCodePro'] = df['YearsCodePro'].replace('Less than 1 year', '0')
df['YearsCodePro'] = df['YearsCodePro'].replace('More than 50 years', '50')
df['YearsCodePro'] = df['YearsCodePro'].astype("int")

bins=[0,4,10,100]
group_names = ['Junior', 'Mid-Level', 'Senior']

#bins

df['ExperienceLevel'] = pd.cut(df['YearsCodePro'], bins, labels=group_names, include_lowest=True)
df['ExperienceLevel'].value_counts()
```
